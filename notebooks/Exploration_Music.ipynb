```python
import pandas as pd
import numpy as np
```

```python
# Example file paths (change these)
music_path = "../data/raw/music/artists.csv"

music = pd.read_csv(music_path)

print(music.head())
```

```python
print(music["tags_lastfm"].head())

print(music["listeners_lastfm"].describe())

print(music["scrobbles_lastfm"].describe())

print(music["ambiguous_artist"].value_counts())
```

```python
music = music[
    (~music["ambiguous_artist"])
]

music = music[
    music["listeners_lastfm"] >= 100
]

print(len(music))
```

```python
print((music["listeners_lastfm"] >= 100).sum())
print((music["listeners_lastfm"] >= 500).sum())
print((music["listeners_lastfm"] >= 1000).sum())
print((music["listeners_lastfm"] >= 5000).sum())
```
