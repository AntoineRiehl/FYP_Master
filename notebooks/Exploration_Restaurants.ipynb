```python
from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path.cwd().parent

DATA_DIR = ROOT / "data" / "raw" / "restaurants"
```

```python
business = pd.read_json(
    DATA_DIR / "business.json",
    lines=True
)

review = pd.read_json(
    DATA_DIR / "review.json",
    lines=True
)

tip = pd.read_json(
    DATA_DIR / "tip.json",
    lines=True
)
```

```python
print("Businesses :", len(business))
print("Reviews    :", len(review))
print("Tips       :", len(tip))
```

```python
print("Business columns")
print(business.columns.tolist())

print("\nReview columns")
print(review.columns.tolist())

print("\nTip columns")
print(tip.columns.tolist())
```
